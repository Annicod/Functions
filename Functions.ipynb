2. Explain the concept of function arguments and parameters in Python.
    - In Python, parameters are the variable names listed in a function definition, while arguments are the actual values supplied to the function when it is called. Parameters act as placeholders for the data the function will receive, and arguments are the real input values given during the function call. Python supports several types of arguments. Positional arguments are matched to parameters based on their order. Keyword arguments are specified by explicitly naming the parameter, allowing you to pass them in any order. Default arguments are parameters that assume a default value if no argument is provided.
    -Example:
     - Here, 'a' and 'b' are parameters.
             def add(a, b):
               return a + b

     - Here, 5 and 3 are arguments.
               result = add(5, 3)
                 print(result)  # Output: 8

6. Explain the concept of generators in Python and how they are defined.
   - A generator in Python is a special type of iterator that allows to iterate over data lazily—that is, one value at a time and only when needed. Generators are used to handle large datasets efficiently or infinite sequences because they do not store all values in memory, unlike lists.
    - Example:
            # Define a generator function to count from 1 to n
            def count_up_to(n):
                count = 1  # Start counting from 1
                while count <= n:
                    yield count  # Yield the current count (pause and return value)
                    count += 1    # Increment count for next iteration
           # Create a generator object
           counter = count_up_to(5)

          # Use a for loop to iterate over the generator
           for num in counter:
              print(num)  # Output: 1, 2, 3, 4, 5 (one by one)




7. What are the advantages of using generators over regular functions?
   - Generators in Python are more memory-efficient than regular functions because they produce values one at a time using the yield keyword instead of storing all results in memory. This makes them ideal for working with large or infinite sequences. They also use lazy evaluation, meaning values are generated only when needed, improving performance. Generators simplify code and help avoid memory overload, while regular functions return all results at once, which can slow down the program for large data.
   - Example:
            # Generator function
            def count_up_to(n):
                count = 1
                while count <= n:
                    yield count  # yields one value at a time
                    count += 1
          # Using the generator
          for num in count_up_to(5):
               print(num)

9. Explain the purpose and usage of the map() function in Python.

10. What is the difference between map(), reduce(), and filter() functions in Python?
   - The map() function used to transform each item in an iterable by applying the specified function whether the reduce() function reduces an iterable to a single cumulative value by applying the function cumulatively and the filter() function used for filtering elements from an iterable based on a condition defined by the function.
    - Example:
             # List of numbers
              nums = [1, 2, 3, 4]

             # Use map() to square each number in the list
              squared = list(map(lambda x: x * x, nums))

             # Use filter() to select even numbers only
              even = list(filter(lambda x: x % 2 == 0, nums))

             # Use reduce() to sum all the numbers in the list
              total = reduce(lambda x, y: x + y, nums)

             # Output:
               print(squared)  # Output: [1, 4, 9, 16]
               print(even)  # Output: [2, 4]
               print(total)  # Output: 10

```python
#1. Write a Python function that takes a list of numbers as input and returns the sum of all even numbers in the list.
def sum_even_numbers(a): # Function to sum even numbers in a list

  sum = 0  # Initialize sum
  for i in a:
    if i%2 == 0:  # Check if the number is even
      sum = sum + i # Add even number to sum
  return sum #return sum

# Example:
a = [1,4,5,6]
print(sum_even_numbers(a)) # Prints sum of even numbers of the list.
```

```python
# 4. Write a Python function that checks if a given number is prime or not from 1 to 200.
# Function to check if a number is prime
def is_prime(n):
    if n < 2:
        return False  # 0 and 1 are not prime
    for i in range(2, n):
        if n % i == 0:
            return False  # If divisible, not prime
    return True  # No divisors found, it's prime


for num in range(1, 201):
    if is_prime(num):
        print(f"{num} is a prime number")
```

```python
#5. Create an iterator class in Python that generates the Fibonacci sequence up to a specified number of terms.
# Generator function to generate Fibonacci sequence
def fibonacci_generator(n):
    a, b = 0, 1  # Starting values
    for _ in range(n):
        yield a  # Yield current value
        a, b = b, a + b  # Update to next pair


for num in fibonacci_generator(10):
    print(num)
```

```python
#6. Write a generator function in Python that yields the powers of 2 up to a given exponent.
# Generator function to yield powers of 2
def powers_of_two(exponent):
    for exp in range(exponent + 1):
        yield 2 ** exp  # Yield 2 raised to the current exponent


for num in powers_of_two(5):
    print(num)
```

```python
# 7. Implement a generator function that reads a file line by line and yields each line as a string.
with open('file.txt', 'w') as file: # Create a file named file.txt.
    file.write("Hello, this is line 1.\n")
    file.write("This is line 2.\n")
    file.write("And this is line 3.\n")
# Generator function to read a file line by line
def read_file_lines(file):
    with open(file, 'r') as file:
        for line in file:
            yield line.strip()  # Yield each line without trailing newline
for line in read_file_lines('file.txt'):
    print(line)
```

```python
#9. Write a Python program that uses `map()` to convert a list of temperatures from Celsius to Fahrenheit.
# Function to convert Celsius to Fahrenheit
def celsius_to_fahren(cel):
    return (cel * 9/5) + 32


celsius_temp = [0, 20, 30, 40, 100]  # List of temperatures in Celsius


fahrenheit_temperatures = list(map(celsius_to_fahren, celsius_temp))  # Use map() to convert the temperatures


print(fahrenheit_temperatures)  # Print the result
```

```python
# 10. Create a Python program that uses `filter()` to remove all the vowels from a given string.

def remove_vowels(char): # Function to filter out vowels
    vowels = "aeiouAEIOU"  # Define vowels
    return char not in vowels  # Return True if the character is not a vowel

input_string = "Hello, this is a test string."

filtered_string = ''.join(filter(remove_vowels, input_string))

# Print the result
print(filtered_string)
```
